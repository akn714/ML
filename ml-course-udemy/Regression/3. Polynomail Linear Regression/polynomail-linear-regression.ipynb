```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv('Position_Salaries.csv')
X = dataset.iloc[:, 1:-1].values
y = dataset.iloc[:, -1:].values
```

```python
# training simple linear regressoin model
from sklearn.linear_model import LinearRegression
lin_reg = LinearRegression()
lin_reg.fit(X, y)
```

```python
# training polynomial linear regression
from sklearn.preprocessing import PolynomialFeatures
poly_reg = PolynomialFeatures(degree=4)
X_poly = poly_reg.fit_transform(X)

lin_reg2 = LinearRegression()
lin_reg2.fit(X_poly, y)
```

```python
# visualising simple linear regressoin model
plt.scatter(X, y, color='red')
plt.plot(X, lin_reg.predict(X), color='blue')
plt.title('Linear Regression (salaries)')
plt.xlabel('position level')
plt.ylabel('salary')
plt.show()
```

```python
# visualising polynomial linear regressoin model
plt.scatter(X, y, color='red')
plt.plot(X, lin_reg2.predict(X_poly), color='blue')
plt.title('Polynomial Linear Regression (salaries)')
plt.xlabel('position level')
plt.ylabel('salary')
plt.show()
```

```python
# visualising polynomial linear regression (higher resolution)
X_grid = np.arange(min(X), max(X), 0.1)
X_grid = X_grid.reshape((len(X_grid), 1))
plt.scatter(X, y, color='red')
plt.plot(X_grid, lin_reg2.predict(poly_reg.fit_transform(X_grid)), color='blue')
plt.title('Polynomial Linear Regression (salaries)')
plt.xlabel('position level')
plt.ylabel('salary')
plt.show()
```

```python
# predicting a new result with linear regression
lin_reg.predict([[6.5]]) # 6.5 -> position level
```

```python
# predicting a new result with polynomial regression
lin_reg2.predict(poly_reg.fit_transform([[6.5]]))
```
