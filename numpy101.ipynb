```python
# right way of copying the arrays
g = np.array([1,2,3])

h = g
h[0] = 100
print(g) # [100, 2, 3]
print(h) # [100, 2, 3]
h[0] = 1

h = g.copy()
h[0] = 100
print(g) # [1, 2, 3] -> original array
print(h) # [100, 2, 3] -> copied array
```

#### Linear Algebra (https://docs.scipy.org/doc/numpy/reference/routines.linalg.html)

# Reorganizing Arrays

```python
ra = np.array(([[1,2,3,4],[5,6,7,8]]))
```

```python
ra2 = np.reshape(ra, (2,2,2))
```

```python
print(ra)
print(ra2)
```

```python
# Vertical stacking of vectors
v1 = np.array(([1,2,3,4]))
v2 = np.array(([5,6,7,8]))
np.vstack([v1, v2, v2, v2, v1])
```

```python
# Horizontal stacking of vectors
h1 = np.ones((2,4))
h2 = np.zeros((2,2))
np.hstack([h1, h2])
```

```python
# check if any value element satisfies any condition
checkCondition = np.array(([[6,2,2,3,6,7], [1,3,6,8,2,3]]))
print(np.any(checkCondition>5, axis=1))
print(np.all(checkCondition>5, axis=0))
checkCondition[np.any(checkCondition>5, axis=1)]
```
